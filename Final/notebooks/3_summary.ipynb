# Tổng kết

### 1. Lý do lựa chọn đề tài
Đầu tiên, nhóm em quyết định lựa chọn bộ dataset này trong bối cảnh trí tuệ nhân tạo đang trở thành một xu hướng bùng nổ trong ngành công nghệ thông tin vào năm 2025. Việc phân tích bộ dữ liệu này không chỉ mang tính thời sự mà còn giúp nhóm đi tìm câu trả lời cho bài toán cốt lõi: **Liệu sự kết hợp giữa con người và AI thực sự tác động như thế nào đến hiệu suất, chất lượng code và sức khỏe tinh thần của lập trình viên?**

### 2. Các key insights thông qua dataset
Tiếp theo, thông qua quá trình đặt câu hỏi, trực quan hóa và kiểm định thống kê, nhóm em đã rút ra được những kết luận sau dựa trên bộ dataset đã chọn:

*   **Cân bằng trong sử dụng AI là vô cùng quan trọng:** Việc phụ thuộc quá ít vào AI khiến lập trình viên mắc nhiều lỗi cú pháp/logic thủ công, nhưng việc lạm dụng AI lại làm giảm hiệu suất do mất thời gian sửa lỗi do AI tạo ra. Mặc dù nhóm sử dụng AI nhiều lại có tỷ lệ hoàn thành task cao nhất, nhưng tỉ lệ gặp lỗi không mong muốn cũng là cao nhất. **Kết luận:** Nên dùng AI để giải quyết vấn đề khó, nhưng phải kết hợp review code kỹ càng thay vì để AI làm hết.

*   **Làm việc thông minh tốt hơn làm việc chăm chỉ:** Kết quả K-means clustering chỉ ra rằng nhóm lập trình viên thành công nhất không phải là nhóm code chay nhiều nhất, mà là nhóm ngủ đủ giấc, stress thấp và tận dụng AI nhiều. Ngược lại, nhóm code thủ công nhiều giờ chỉ đạt kết quả trung bình, và nhóm bị stress cao luôn có hiệu suất thấp nhất.

*   **Tải nhận thức là rào cản lớn nhất:** Phân tích SHAP value cho thấy tải nhận thức có tác động tiêu cực mạnh nhất đến sự thành công của dự án, trong khi số giờ ngủ hay lượng cà phê uống có tác động không đáng kể. Điều này gợi ý rằng áp lực không đơn thuần đến từ việc thiếu ngủ, mà đến từ sự quản lý công việc và sức khỏe tinh thần.

*   **Cà phê không giúp tăng hiệu suất:** Việc nạp nhiều cà phê không giúp giảm stress hay cải thiện chất lượng code, ngược lại còn khiến số lượng bug sinh ra nhiều hơn (đặc biệt ở nhóm tiêu thụ trung bình) so với nhóm uống ít nhưng tỉnh táo.

*   **Tâm lý ổn định giúp cải thiện việc sử dụng AI:** Khi so sánh cùng một lượng thời gian dùng AI, nhóm có tinh thần ổn định chuyển hóa công cụ thành hiệu quả công việc tốt hơn **~26%** so với nhóm đang bị quá tải về mặt tinh thần.

### 3. Kiến thức & kỹ năng học được
Thông qua project này, nhóm em đã học được nhiều thứ:
*   Thành thạo hơn các thư viện: **numpy, pandas** để xử lí dữ liệu; **matplotlib, seaborn** để trực quan hóa.
*   Học & áp dụng các mô hình Machine Learning với **Scikit-learn**: K-Means Clustering, Random Forest, Linear Regression.
*   Sử dụng kiểm định thống kê **ANOVA** để đánh giá sự khác biệt có ý nghĩa giữa các nhóm dữ liệu.

### 4. Hướng đi tương lai của dự án
*   Thay vì chỉ tìm tương quan, chúng em có thể xác định rõ xem stress gây ra việc uống nhiều cà phê hay ngược lại.
*   Có thể mở rộng dự án bằng cách tạo một ứng dụng để trực quan hóa các chỉ số sức khỏe và năng suất cho các quản lý dự án hoặc tạo một công cụ gợi ý lịch nghỉ ngơi hoặc cảnh báo khi các hiệu ứng tiêu cực đạt ngưỡng gây hại cho hiệu suất.
*   Sử dụng các thuật toán phân cụm khác hoặc sử dụng các phương pháp chọn số lượng cụm cho các thuật toán phân cụm thay vì gán cứng như đã làm.
