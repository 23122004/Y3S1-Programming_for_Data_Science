# Pandas Hands-on Exercise - California Baby Names

### Dataset Introduction
This dataset (attached) contains information about **baby names** registered in California (USA) from 1910 to 2022.

#### Data Structure:
- **State**: State (CA = California)
- **Sex**: Gender (F = Female, M = Male)
- **Year**: Year of birth
- **Name**: Name
- **Count**: The number of babies with that name in that year

#### Example:
```
CA,F,2020,Olivia,2353  → In 2020, there were 2,353 baby girls named Olivia
CA,M,2020,Noah,2458    → In 2020, there were 2,458 baby boys named Noah
```

### Setup

````python
import pandas as pd
import numpy as np
````

````python
# Read the DataFrame from a CSV file, with the following columns
field_names = ['State', 'Sex', 'Year', 'Name', 'Count']
babynames_df = pd.read_csv('Data/STATE.CA.TXT',header=None,names=field_names)
````

### Task 2: Trend Analysis: analyzing how the Name "Emma" change over time
**Instruction**
- Get all records for the name "Emma"
- In which year did Emma first appear?
- In which year was Emma most popular
- How many baby boys were named Emma?
- Calculate the total number of babies named Emma from 1910 to 2022.

````python
# Task 2: Trend Analysis for "Emma"
print("=" * 60)
print("TASK 2: Trend Analysis for 'Emma'")
print("=" * 60)

# Get all records for Emma
emma_df = babynames_df[babynames_df['Name'] == 'Emma']

print(f"\n1. All records for Emma:")
print(f"   - Total records: {len(emma_df)}")

# First year Emma appeared
print(f"\n2. First year Emma appeared: {emma_df['Year'].min()}")

# Most popular year
most_popular_year = emma_df.groupby('Year')['Count'].sum().idxmax()
most_popular_count = emma_df.groupby('Year')['Count'].sum().max()
print(f"\n3. Most popular year: {most_popular_year} (with {most_popular_count:,} babies)")

# Baby boys named Emma
boys_named_emma = emma_df[emma_df['Sex'] == 'M']['Count'].sum()
print(f"\n4. Baby boys named Emma: {boys_named_emma}")

# Total number of babies named Emma
total_emma = emma_df['Count'].sum()
print(f"\n5. Total babies named Emma (1910-2022): {total_emma:,}")

print("\n" + "=" * 60)
````

### Task 3: Analysis by Decade (1910s, 1920s, ..., 2020s)
**Instruction**
- Calculate the total number of babies born in each decade (create Decade column).
- Which decade had the highest number of babies?
- In the 1990s decade, how many unique names were there?
- "Very Popular" name mean Count >= 500. Compare the number of "Very Popular" names between the 1980s and 2010s decades.
- Calculate the average Count for each decade.

````python
# Task 3: Analysis by Decade
print("=" * 60)
print("TASK 3: Analysis by Decade")
print("=" * 60)

# Create Decade column
babynames_df['Decade'] = (babynames_df['Year'] // 10) * 10

# Total babies per decade
babies_per_decade = babynames_df.groupby('Decade')['Count'].sum()
print(f"\n1. Total babies born in each decade:")
print(babies_per_decade)

# Decade with highest number of babies
max_decade = babies_per_decade.idxmax()
max_count = babies_per_decade.max()
print(f"\n2. Decade with highest number of babies: {max_decade}s (with {max_count:,} babies)")

# Unique names in 1990s
names_1990s = babynames_df[babynames_df['Decade'] == 1990]['Name'].nunique()
print(f"\n3. Unique names in 1990s: {names_1990s:,}")

# Very Popular names (Count >= 500) comparison
very_popular_1980s = babynames_df[(babynames_df['Decade'] == 1980) & (babynames_df['Count'] >= 500)].shape[0]
very_popular_2010s = babynames_df[(babynames_df['Decade'] == 2010) & (babynames_df['Count'] >= 500)].shape[0]
print(f"\n4. 'Very Popular' names comparison:")
print(f"   - 1980s: {very_popular_1980s:,} names with Count >= 500")
print(f"   - 2010s: {very_popular_2010s:,} names with Count >= 500")
print(f"   - Difference: {abs(very_popular_1980s - very_popular_2010s):,}")

# Average Count per decade
avg_count_per_decade = babynames_df.groupby('Decade')['Count'].mean()
print(f"\n5. Average Count for each decade:")
print(avg_count_per_decade.round(2))

print("\n" + "=" * 60)
````

### Task 4: Analyze `Timeless` Names (Unchanging Over Time)
**Definition**: A name is called `timeless` if it:
- Appeared in **ALL** decades from the 1910s to the 2020s
- Always had a Count $\ge$ 100 in every decade

**Instruction**
- Find all "timeless" names.
- How many such names are there?
- Display the list of these names.
- Among these names, which one has the highest total Count?

````python
# Task 4: Analyze Timeless Names
print("=" * 60)
print("TASK 4: Timeless Names Analysis")
print("=" * 60)

# Make sure Decade column exists (it was created in Task 3)
if 'Decade' not in babynames_df.columns:
    babynames_df['Decade'] = (babynames_df['Year'] // 10) * 10

# Get all decades from 1910s to 2020s
all_decades = [1910, 1920, 1930, 1940, 1950, 1960, 1970, 1980, 1990, 2000, 2010, 2020]
num_decades = len(all_decades)

# For each name, check if it appears in all decades with Count >= 100
# First, get sum of counts per name per decade
name_decade_counts = babynames_df.groupby(['Name', 'Decade'])['Count'].sum().reset_index()

# Filter only rows with Count >= 100
name_decade_counts_filtered = name_decade_counts[name_decade_counts['Count'] >= 100]

# Count how many decades each name appears in (with Count >= 100)
name_decade_appearance = name_decade_counts_filtered.groupby('Name')['Decade'].count().reset_index()
name_decade_appearance.columns = ['Name', 'DecadeCount']

# Timeless names are those that appear in all decades
timeless_names = name_decade_appearance[name_decade_appearance['DecadeCount'] == num_decades]['Name'].tolist()

print(f"\n1. Number of timeless names: {len(timeless_names)}")

print(f"\n2. List of timeless names:")
print(sorted(timeless_names))

# Find which timeless name has the highest total count
if len(timeless_names) > 0:
    timeless_df = babynames_df[babynames_df['Name'].isin(timeless_names)]
    total_counts = timeless_df.groupby('Name')['Count'].sum().sort_values(ascending=False)
    top_name = total_counts.idxmax()
    top_count = total_counts.max()

    print(f"\n3. Timeless name with highest total Count:")
    print(f"   - Name: {top_name}")
    print(f"   - Total Count: {top_count:,}")
    print(f"\n   Top 10 timeless names by total count:")
    print(total_counts.head(10))

print("\n" + "=" * 60)
````

### Task 5: `Your Name` Analysis
**Perform an in-depth analysis of any given name**\
Create a function `analyze_name(name)` that performs the following:

1. Check if the name exists in the dataset.
2. If it exists, display:
   - The first year it appeared.
   - The most popular year (and its Count).
   - The total number of people with this name.
   - Which gender uses this name more frequently.
   - Which category this name belongs to (Rare/Uncommon/Common/Very Popular) in the last 5 years.
     
***Test with these name:***
```
['Emma', 'Noah', 'Michael', 'Sophia', 'James']
```

````python
# Task 5: Your Name Analysis Function
def analyze_name(name):
    """
    Perform an in-depth analysis of a given name.
    """
    print("=" * 60)
    print(f"NAME ANALYSIS: {name}")
    print("=" * 60)

    # Check if name exists in dataset
    name_df = babynames_df[babynames_df['Name'] == name]

    if len(name_df) == 0:
        print(f"\n❌ The name '{name}' does not exist in the dataset.")
        print("=" * 60)
        return

    print(f"\n✓ The name '{name}' exists in the dataset!")

    # 1. First year appeared
    first_year = name_df['Year'].min()
    print(f"\n1. First year appeared: {first_year}")

    # 2. Most popular year and its count
    year_totals = name_df.groupby('Year')['Count'].sum()
    most_popular_year = year_totals.idxmax()
    most_popular_count = year_totals.max()
    print(f"\n2. Most popular year: {most_popular_year}")
    print(f"   - Count in that year: {most_popular_count:,}")

    # 3. Total number of people with this name
    total_count = name_df['Count'].sum()
    print(f"\n3. Total number of people with this name: {total_count:,}")

    # 4. Which gender uses this name more frequently
    gender_counts = name_df.groupby('Sex')['Count'].sum()
    if len(gender_counts) > 1:
        more_frequent_gender = gender_counts.idxmax()
        gender_name = 'Female' if more_frequent_gender == 'F' else 'Male'
        print(f"\n4. Gender distribution:")
        print(f"   - Female: {gender_counts.get('F', 0):,} ({gender_counts.get('F', 0) / total_count * 100:.1f}%)")
        print(f"   - Male: {gender_counts.get('M', 0):,} ({gender_counts.get('M', 0) / total_count * 100:.1f}%)")
        print(f"   - More frequently used by: {gender_name}")
    else:
        gender = gender_counts.index[0]
        gender_name = 'Female' if gender == 'F' else 'Male'
        print(f"\n4. Gender: {gender_name} only (100%)")

    # 5. Category in the last 5 years (2018-2022)
    max_year = babynames_df['Year'].max()
    last_5_years = name_df[name_df['Year'] >= (max_year - 4)]

    if len(last_5_years) > 0:
        avg_count_last_5 = last_5_years['Count'].mean()

        # Categorize based on average count
        if avg_count_last_5 < 10:
            category = "Rare"
        elif avg_count_last_5 < 100:
            category = "Uncommon"
        elif avg_count_last_5 < 500:
            category = "Common"
        else:
            category = "Very Popular"

        print(f"\n5. Category in the last 5 years ({max_year - 4}-{max_year}):")
        print(f"   - Average Count: {avg_count_last_5:.1f}")
        print(f"   - Category: {category}")
    else:
        print(f"\n5. No data available for the last 5 years.")

    print("\n" + "=" * 60)

# Test with the given names
test_names = ['Emma', 'Noah', 'Michael', 'Sophia', 'James']

print("\n\n" + "=" * 60)
print("TESTING NAME ANALYSIS FUNCTION")
print("=" * 60 + "\n")

for name in test_names:
    analyze_name(name)
    print("\n")
````
