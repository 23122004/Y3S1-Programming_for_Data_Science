## Tiền xử lý (0.5đ)

Ở đây, bạn sẽ chuyển cột "AuthorDate" từ dạng chuỗi sang datetime. <font color=gray>Gợi ý: bạn xem [document](https://docs.python.org/3/library/datetime.html#strftime-and-strptime-format-codes) để biết cách truyền format phù hợp cho hàm `strptime`.</font>

````python
# YOUR CODE HERE
from datetime import datetime, timezone, timedelta

author_date_col = cols.get('AuthorDate', [])
converted_dates = []
for date_str in author_date_col:
    # Tách chuỗi ngày và múi giờ
    parts = date_str.split()
    date_part = ' '.join(parts[1:4])
    time_part = parts[4]
    tz_part = parts[5]

    # Chuyển đổi tháng từ dạng chữ sang số
    month_map = {'Jan': 1, 'Feb': 2, 'Mar': 3, 'Apr': 4, 'May': 5, 'Jun': 6, 
                 'Jul': 7, 'Aug': 8, 'Sep': 9, 'Oct': 10, 'Nov': 11, 'Dec': 12}
    day, month_str, year = parts[1], parts[2], parts[3]
    month = month_map[month_str]
    
    # Tạo chuỗi mới có thể parse được
    parseable_date_str = f'{day} {month} {year} {time_part}'
    dt_obj = datetime.strptime(parseable_date_str, '%d %m %Y %H:%M:%S')
    
    # Xử lý múi giờ
    if tz_part.startswith(('+', '-')):
        sign = 1 if tz_part[0] == '+' else -1
        hours = int(tz_part[1:3])
        minutes = int(tz_part[3:5])
        tz_offset = timedelta(hours=hours, minutes=minutes)
        if sign == -1:
            tz_offset = -tz_offset
    else:
        tz_offset = timedelta(0) # Mặc định là UTC nếu không có thông tin
        
    converted_dates.append(dt_obj.replace(tzinfo=timezone(tz_offset)))

cols['AuthorDate'] = converted_dates
````

````python
# TEST
assert type(cols['AuthorDate'][0]) == datetime.datetime
assert cols['AuthorDate'][0] == datetime.datetime(2025, 10, 9, 11, 32, 46, 
            tzinfo=datetime.timezone(datetime.timedelta(seconds=7200)))
````

### Với mỗi cột có kiểu dữ liệu dạng phân loại (categorical), các giá trị được phân bố như thế nào? (1.75đ)

Các cột dạng phân loại là các cột mà không phải là cột "AuthorDate". Bạn sẽ lưu thông tin của các cột này vào biến `cate_col_profiles`. Biến này là một dictionary, trong đó mỗi phần tử ứng với một cột dạng phân loại, key là tên cột, value là list gồm: tỉ lệ % giá trị thiếu, số lượng các giá trị khác nhau (không tính giá trị thiếu), set chứa các giá trị khác nhau (không bao gồm giá trị thiếu).

````python
# YOUR CODE HERE
cate_col_profiles = {}
categorical_cols = [col for col in cols if col != 'AuthorDate']
total_rows = len(cols['Id'])

for col_name in categorical_cols:
    col_values = cols[col_name]
    non_empty_values = [v for v in col_values if v != '']
    
    missing_percentage = (1 - len(non_empty_values) / total_rows) * 100
    distinct_values = set(non_empty_values)
    num_distinct_values = len(distinct_values)
    
    cate_col_profiles[col_name] = [missing_percentage, num_distinct_values, distinct_values]
````

## Tiền xử lý (nếu cần) + phân tích dữ liệu để trả lời cho câu hỏi 1 (1đ)

Nhắc lại câu hỏi 1: *trong năm 2023, ai là người có nhiều commit nhất, ai là người có nhiều commit nhì, ...?*

Bạn sẽ lưu kết quả vào biến `wanted_people`. Biến này là một dictionary, mỗi phần tử ứng với một người mà có commit trong năm 2023, key là tên của người đó, value là số lượng commit của người đó. Dictionary này được sort theo value giảm dần.

````python
# YOUR CODE HERE
from collections import defaultdict

commits_2023 = defaultdict(int)
for i in range(len(cols['Id'])):
    if cols['AuthorDate'][i].year == 2023:
        author_name = cols['AuthorName'][i]
        commits_2023[author_name] += 1

wanted_people = dict(sorted(commits_2023.items(), key=lambda item: item[1], reverse=True))
````

## Tiền xử lý (nếu cần) + phân tích dữ liệu để trả lời cho câu hỏi 2 (2đ)

Nhắc lại câu hỏi 2: *commit nào là commit cuối cùng (xét trên "AuthorDate") đã thay đổi file "nbdime/webapp/templates/difftool.html"?*

Ta thống nhất: không xét các commit "merge pull request" vì các commit này không có thông tin về "ChangedFiles".

Bạn lưu kết quả vào biến `wanted_commit`. Biến này là một dictionary, trong đó mỗi phần tử ứng với một thông tin của commit, key là tên cột, value là giá trị tương ứng.

````python
# YOUR CODE HERE
target_file = 'nbdime/webapp/templates/difftool.html'
latest_commit_index = -1
latest_date = None

for i in range(len(cols['Id'])):
    # Bỏ qua các merge commit (không có thông tin file thay đổi)
    if not cols['ChangedFiles'][i]:
        continue
    
    if target_file in cols['ChangedFiles'][i].split(','):
        current_date = cols['AuthorDate'][i]
        if latest_date is None or current_date > latest_date:
            latest_date = current_date
            latest_commit_index = i

wanted_commit = {}
if latest_commit_index != -1:
    for col_name in cols:
        wanted_commit[col_name] = cols[col_name][latest_commit_index]

wanted_commit
````
