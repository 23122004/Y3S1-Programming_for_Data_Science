## Tiền xử lý + phân tích dữ liệu để trả lời cho câu hỏi 1

*Câu hỏi 1: Cột điểm tổng “Grade/10.00” có phân bố các giá trị như thế nào?*

Ở đây không cần tiền xử lý mà sẽ đi luôn vào phân tích dữ liệu.

Đầu tiên, ta sẽ xem thử median của cột "Grade/10.00".

```python
grades = cols['Grade/10.00']
sorted_grades = sorted(grades)
p = 50
i = int(p/100 * (len(sorted_grades)-1))
sorted_grades[i]
```

Để cảm thụ rõ hơn về phân bố các giá trị của cột "Grade/10.00", ta sẽ chia miền giá trị của cột này thành các khoảng (bin) và xem thử mỗi khoảng có bao nhiêu sinh viên.

Giả sử ta sẽ chia đoạn [0, 10] thành 5 bin như sau: [0, 2), [2, 4), [4, 6), [6, 8),[8,10]. Ta có thể lưu kết quả vào một dictionary, trong đó: key là một tuple cho biết 2 đầu của một bin (thay tuple bằng list thì có được không?), value là số lượng sinh viên rơi vào bin này.

***Sinh viên hoàn thành tiếp phần này như bài tập về nhà***

Thay `tuple` bằng `list` thì có được không? Trả lời trực tiếp vào cell này

```python
print('Không thay tuple bằng list được vì list là kiểu dữ liệu mutable, nên không thể dùng làm key trong dictionary được. Còn tuple là kiểu dữ liệu immutable, nên có thể dùng làm key trong dictionary.')

grades = cols['Grade/10.00']

bins = {
    (0, 2): 0,
    (2, 4): 0,
    (4, 6): 0,
    (6, 8): 0,
    (8, 10): 0
}

for grade in grades:
    if 0 <= grade < 2:
        bins[(0, 2)] += 1
    elif 2 <= grade < 4:
        bins[(2, 4)] += 1
    elif 4 <= grade < 6:
        bins[(4, 6)] += 1
    elif 6 <= grade < 8:
        bins[(6, 8)] += 1
    elif 8 <= grade <= 10:
        bins[(8, 10)] += 1

bins
```

## Câu hỏi 3
**Có mối tương quan giữa thời gian làm bài và điểm số không?**\
**Mục đích**: Tìm hiểu xem sinh viên làm bài nhanh hay chậm có ảnh hưởng đến điểm số không.

### Đơn giản
#### Bước 1: Tính thống kê cơ bản
- Tính thời gian trung bình của tất cả sinh viên
- Tính thời gian trung vị (median) - đây là mốc chia sinh viên thành 2 nhóm

#### Bước 2: Chia nhóm so sánh
- **Nhóm làm nhanh:** Sinh viên làm bài dưới thời gian trung vị
- **Nhóm làm chậm:** Sinh viên làm bài trên hoặc bằng thời gian trung vị
- Tính điểm trung bình của mỗi nhóm

#### Bước 3: Phân tích kết quả

**Có 3 tình huống có thể xảy ra:**

- **Nhóm làm nhanh có điểm cao hơn:** Có thể sinh viên giỏi nắm kiến thức vững nên làm nhanh và chính xác
- **Nhóm làm chậm có điểm cao hơn:** Có thể sinh viên cẩn thận, suy nghĩ kỹ sẽ đạt kết quả tốt hơn
- **Hai nhóm điểm tương đương:** Không có mối liên hệ rõ ràng giữa thời gian và điểm số

***Sinh viên hoàn thành tiếp trong các cell kế như bài tập về nhà***

## Bước 1: Tính thống kê cơ bản

```python
time_taken = cols['Time taken']
grades = cols['Grade/10.00']

sorted_time = sorted(time_taken)
num_students = len(sorted_time)
median_index = num_students // 2
median_time = sorted_time[median_index]

print(f"Thời gian trung vị: {median_time:.2f} phút.")
```

## Bước 2: Chia nhóm so sánh

```python
fast_group = []
slow_group = []

for i in range(len(time_taken)):
    if time_taken[i] < median_time:
        fast_group.append(grades[i])
    else:
        slow_group.append(grades[i])

avg_group_fast = sum(fast_group) / len(fast_group) if fast_group else 0
avg_group_slow = sum(slow_group) / len(slow_group) if slow_group else 0

print(f"Điểm trung bình nhóm làm nhanh: {avg_group_fast:.2f}")
print(f"Điểm trung bình nhóm làm chậm: {avg_group_slow:.2f}")
```

## Bước 3: Phân tích kết quả

Vì điểm trung bình của nhóm làm chậm cao hơn nên suy ra được sinh viên cẩn thận, suy nghĩ kỹ có thể sẽ đạt kết quả tốt hơn.

### Nâng cao (không cần thư viện)
- Chia thời gian thành nhiều khoảng (ví dụ: 0-10 phút, 10-20 phút, 20-30 phút...)
- Tính điểm trung bình cho mỗi khoảng thời gian
- Quan sát xu hướng: điểm có tăng hay giảm theo thời gian không?

***Sinh viên hoàn thành tiếp trong các cell kế như bài tập về nhà***

```python
time_bins = {}
max_time = max(time_taken)

for i in range(0, int(max_time) + 10, 10):
    time_bins[(i, i + 10)] = []

for i in range(len(time_taken)):
    for t_range in time_bins:
        if t_range[0] <= time_taken[i] < t_range[1]:
            time_bins[t_range].append(grades[i])
            break

print("Điểm trung bình theo khoảng thời gian:")
for t_range, grade_list in time_bins.items():
    if grade_list:
        avg_grade = sum(grade_list) / len(grade_list)
        print(f"{t_range[0]}-{t_range[1]} phút: {avg_grade:.2f} ({len(grade_list)} sinh viên)")
```

Kết quả cho thấy điểm trung bình có xu hướng **tăng lên** theo thời gian.

## Câu hỏi 4
**Có sinh viên nào là điểm ngoại lệ (outliers) không?**\
**Mục đích:** Tìm những sinh viên có điểm số bất thường (quá cao hoặc quá thấp so với mức trung bình).

### Phương pháp 1: Sử dụng khoảng tứ phân vị (IQR - Interquartile Range)

#### Bước 1: Tính các giá trị thống kê
- Sắp xếp tất cả điểm số từ thấp đến cao
- Tính **Q1** (tứ phân vị thứ nhất): Điểm ở vị trí 25% (1/4 đầu dãy số)
- Tính **Q3** (tứ phân vị thứ ba): Điểm ở vị trí 75% (3/4 dãy số)
- Tính **IQR** = Q3 - Q1 (khoảng giữa)

**Ví dụ:** Có 100 sinh viên sau khi sắp xếp
- Q1 = điểm của sinh viên thứ 25 (ví dụ: 4.5)
- Q3 = điểm của sinh viên thứ 75 (ví dụ: 7.5)
- IQR = 7.5 - 4.5 = 3.0

#### Bước 2: Xác định ngưỡng outlier
- **Ngưỡng dưới** = Q1 - 1.5 × IQR
- **Ngưỡng trên** = Q3 + 1.5 × IQR

Tiếp ví dụ trên:
- Ngưỡng dưới = 4.5 - 1.5 × 3.0 = 0
- Ngưỡng trên = 7.5 + 1.5 × 3.0 = 12 (nhưng tối đa là 10)

#### Bước 3: Tìm outliers
- Sinh viên có điểm < ngưỡng dưới → **Outlier thấp** (điểm kém bất thường)
- Sinh viên có điểm > ngưỡng trên → **Outlier cao** (điểm xuất sắc bất thường)

***Sinh viên hoàn thành tiếp trong các cell kế như bài tập về nhà***

## Bước 1: Tính các giá trị thống kê

```python
grades = cols['Grade/10.00']
sorted_grades = sorted(grades)
n = len(sorted_grades)

q1_index = int((0.25 * n) - 1)
q3_index = int((0.75 * n) - 1)
q1 = sorted_grades[q1_index]
q3 = sorted_grades[q3_index]
iqr = q3 - q1

print(f"Q1: {q1}")
print(f"Q3: {q3}")
print(f"IQR: {iqr:.2f}")
```

## Bước 2: Xác định ngưỡng outlier

```python
lower_bound = q1 - 1.5 * iqr
upper_bound = q3 + 1.5 * iqr

print(f"Ngưỡng dưới: {lower_bound:.2f}")
print(f"Ngưỡng trên: {upper_bound:.2f}")
```

## Bước 3: Tìm outliers

```python
low_outliers = [grade for grade in grades if grade < lower_bound]
high_outliers = [grade for grade in grades if grade > upper_bound]

print(f"Outliers thấp: {low_outliers}")
print(f"Outliers cao: {high_outliers}")
```

### Phương pháp 2: Sử dụng độ lệch chuẩn (Standard Deviation)

#### Bước 1: Tính giá trị trung bình và độ lệch chuẩn
- Tính điểm trung bình (mean)
- Tính độ lệch chuẩn (standard deviation) - đo mức độ phân tán của điểm số

#### Bước 2: Xác định ngưỡng
- **Ngưỡng dưới** = Mean - 2 × Std Dev
- **Ngưỡng trên** = Mean + 2 × Std Dev

**Ví dụ:**
- Điểm trung bình = 6.0
- Độ lệch chuẩn = 1.5
- Ngưỡng dưới = 6.0 - 2 × 1.5 = 3.0
- Ngưỡng trên = 6.0 + 2 × 1.5 = 9.0

#### Bước 3: Tìm outliers
- Điểm < 3.0 hoặc > 9.0 là outliers

***Sinh viên hoàn thành tiếp trong các cell kế như bài tập về nhà***

## Bước 1: Tính giá trị trung bình và độ lệch chuẩn

```python
grades = cols['Grade/10.00']
n = len(grades)

mean = sum(grades) / n

variance = sum([(x - mean) ** 2 for x in grades]) / n
std_dev = variance ** 0.5

print(f"Điểm trung bình: {mean:.2f}")
print(f"Độ lệch chuẩn: {std_dev:.2f}")
```

## Bước 2: Xác định ngưỡng

```python
lower_bound = mean - 2 * std_dev
upper_bound = mean + 2 * std_dev

print(f"Ngưỡng dưới: {lower_bound:.2f}")
print(f"Ngưỡng trên: {upper_bound:.2f}")
```

## Bước 3: Tìm outliers

```python
outliers = [grade for grade in grades if grade < lower_bound or grade > upper_bound]

outliers
```
